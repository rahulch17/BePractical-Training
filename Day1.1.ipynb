```python
num=int(input("Enter a num: "))
if num < 0:
  print("Sorry, factorial does not exist.")
elif num ==1:
  print("The factorial of 0 is 1")
else:
  factorial = 1
  for i in range(1, num + 1):
    factorial *= i
  print(f"The factorial of {num} is {factorial}")
```

```python
#Activity: Exploring List Functins using a fruits list
fruits=["Apple","banana","Mango","Orange"]
print(fruits)

fruits.append("Grapes")
more_fruits=["Pineapple","cherry"]
fruits.extend(more_fruits)
print("After extend",fruits)

print("first fruit: ",fruits[0])
print("Last fruits: ", fruits[-1])
print(fruits[2:5])


fruits[1]="strawberry"
print("after updating index 1: ", fruits)
fruits.remove("Mango")
print("After remove Manfo: ", fruits)

removed_fruits=fruits.pop(2)
print(removed_fruits)
print("After pop: ", fruits)

fruits.sort()
print("After sort: ", fruits)

fruits.sort()
print("After sort: ", fruits)
fruits.reverse()
print("After reverse: ",fruits)

fruits.clear()
print(fruits)
```
